Employee data analysis for ABC Company.

```python
# Module 1 End Project
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_excel('ABC Company.xlsx')
df.head()

# Preprocessing
np.random.seed(42)
df['Height']=np.random.randint(150,181,size=len(df))
df.head()

# 1. Team distribution
team_dist=df['Team'].value_counts()
team_pct=(team_dist/len(df))*100
print(pd.DataFrame({'Count':team_dist,'Percentage':team_pct.round(2)}))
team_dist.plot(kind='bar')
plt.title('Employees by Team')
plt.show()

# 2. Position distribution
pos=df['Position'].value_counts()
print(pos)
pos.plot(kind='bar')
plt.title('Employees by Position')
plt.xticks(rotation=45)
plt.show()

# 3. Predominant age group
bins=[20,25,30,35,40,45,50]
labels=['20-25','26-30','31-35','36-40','41-45','46-50']
df['AgeGroup']=pd.cut(df['Age'],bins=bins,labels=labels,include_lowest=True)
agegrp=df['AgeGroup'].value_counts().sort_index()
print(agegrp)
agegrp.plot(kind='bar')
plt.title('Age Group Distribution')
plt.show()

# 4. Highest salary expenditure
team_salary=df.groupby('Team')['Salary'].sum().sort_values(ascending=False)
pos_salary=df.groupby('Position')['Salary'].sum().sort_values(ascending=False)
print(team_salary.head(1))
print(pos_salary.head(1))
team_salary.plot(kind='bar')
plt.title('Salary by Team')
plt.show()
pos_salary.plot(kind='bar')
plt.title('Salary by Position')
plt.xticks(rotation=45)
plt.show()

# 5. Correlation
corr=df['Age'].corr(df['Salary'])
print('Correlation:',corr)
plt.scatter(df['Age'],df['Salary'])
plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Age vs Salary')
plt.show()

print('Insights:')
print('- Team distribution varies across departments.')
print('- Employee positions are unevenly distributed.')
print('- One age group has the highest employee count.')
print('- Salary expenditure differs by team and position.')
print('- Correlation value indicates the relationship between age and salary.')
```
