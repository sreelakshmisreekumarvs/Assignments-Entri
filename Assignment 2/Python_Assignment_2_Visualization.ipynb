## Visualization with Seaborn and Matplotlib

**Dataset:** Titanic dataset from Seaborn


### Import Libraries and Load Dataset


```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset
titanic = sns.load_dataset('titanic')

# Display first 5 rows
titanic.head()
```

## 1. Countplot: Passenger Count by Embarked Port


```python
plt.figure(figsize=(8, 5))

sns.countplot(data=titanic, x='embarked', hue='sex')

plt.title('Passenger Count by Embarked Port')
plt.xlabel('Embarked Port')
plt.ylabel('Number of Passengers')
plt.legend(title='Sex')

plt.show()
```

**Observation:**  
Most passengers embarked from port S (Southampton). The number of male passengers is higher than female passengers, especially at Southampton.


## 2. Swarmplot: Age Distribution across Classes


```python
plt.figure(figsize=(8, 5))

sns.swarmplot(data=titanic, x='pclass', y='age', hue='sex', size=3)

plt.title('Age Distribution across Passenger Classes')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.legend(title='Sex')

plt.show()
```

**Observation:**  
Passengers of different ages are present in all three classes. Third class has a larger concentration of younger passengers.


## 3. Pie Chart: Gender Distribution


```python
gender_counts = titanic['sex'].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%',
    explode=(0.1, 0)
)

plt.title('Gender Distribution of Titanic Passengers')
plt.show()
```

**Observation:**  
The Titanic dataset contains a higher percentage of male passengers than female passengers.


## 4. Heatmap: Correlation Matrix


```python
columns = ['age', 'fare', 'pclass', 'sibsp', 'parch', 'survived']
correlation = titanic[columns].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(correlation, annot=True)

plt.title('Correlation Matrix of Titanic Features')
plt.show()
```

**Observation:**  
Survival has a negative correlation with passenger class (pclass), while fare shows a positive relationship with survival.


## 5. Violin Plot: Fare by Class and Gender


```python
plt.figure(figsize=(8, 5))

sns.violinplot(data=titanic, x='pclass', y='fare', hue='sex')

plt.title('Fare Distribution by Passenger Class and Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.legend(title='Sex')

plt.show()
```

**Observation:**  
First-class passengers generally paid higher fares compared with second- and third-class passengers.


## 6. Scatter Plot: Age vs Fare


```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=titanic, x='age', y='fare', hue='survived')

plt.title('Age vs Fare by Survival Status')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(title='Survived')

plt.show()
```

**Observation:**  
Passengers who paid higher fares appear to have a greater proportion of survivors, while age alone does not show a clear relationship with fare.
